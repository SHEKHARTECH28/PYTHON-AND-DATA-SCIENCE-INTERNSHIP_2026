```python
# 2) Write a function for basic math operations like add multiply substract divide and use this in your program, take 2 number input from user.

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b != 0:
        return a / b
    else:
        return "Division by zero is not possible"



n1 = float(input("Enter first number: "))
n2 = float(input("Enter second number: "))

print("Addition =", add(n1, n2))
print("Subtraction =", subtract(n1, n2))
print("Multiplication =", multiply(n1, n2))
print("Division =", divide(n1, n2))
```

```python

 #3) Write a program to check Palindrome Number For example Number 12321 is a Palindrome Number, because 12321 is equal to its reverse Number 12321. Steps for checking Palindrome number

 #1. Find reverse of the given number.
 #2. Compare that number with the reverse number.
 #3. If number and its reverse is equal then it is a Palindrome Number otherwise not.

n=input("Enter a number: ")
reverse_n=n[::-1]
if n==reverse_n:
    print(n, "is a palindrome.")
else:
    print(n, "is not a palindrome.")
```

```python
# 1) Practise
# 	Dictionary

print("DICTIONARY")
student={
    "name":"Shekhar",  
    "age": 20,
    "major": "Computer Science"}
print(student)
print("Accessing values:")
print("Name:", student["name"])
print("Age:", student["age"])
print("ADDING A NEW ELEMENT")
student["year"]="2nd_year"
print(student)
print("UPDATING A ELEMENT")
student["age"]=21
print(student)
print("DELETING A ELEMENT")
del student["major"]
print(student)
print("student.keys():", student.keys())
print("student.values():", student.values())
print("student.items():", student.items())



```

```python
#1) Practise
#	Tuple


numbers=(1,23,45,63,94,94,94,95)
print(numbers)
print("Accessing values:")
print("First number:", numbers[0])
print("Slicing:", numbers[1:4])
print("length:", len(numbers))
print("Count of 94:", numbers.count(94))
print("Index of first occurrence of 94:", numbers.index(94))
print("sum:", sum(numbers))

```

```python
# 1) Practise
#	Set

cars={"audi", "bmw", "ferrari", "audi", "gtr"}
print(cars)
print("Adding a car:")
cars.add("nissan")
print(cars)
print("Removing a car:")
cars.remove("gtr")
cars2={"defender", "mercedes", "kia", "audi"}
print("Union:", cars.union(cars2))
print("Intersection:", cars.intersection(cars2))
print("Difference:", cars.difference(cars2))
print("cars in car1 but not in cars2:", cars.difference(cars2))
```
