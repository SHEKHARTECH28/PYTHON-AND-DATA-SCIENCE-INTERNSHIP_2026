```python
arr = np.array([[10, 20, 30],
                [40, 50, 60]])

# Max & Min
print("Max:", arr.max())
print("Min:", arr.min())


print("Shape:", arr.shape)


print("All elements:", arr)
print("Specific element [0,1]:", arr[0,1])

total = 0
for row in arr:
    for val in row:
        total += val
print("Sum using loop:", total)

# Arithmetic operations
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
print("Addition:", arr1 + arr2)
print("Subtraction:", arr1 - arr2)
print("Multiplication:", arr1 * arr2)
print("Division:", arr1 / arr2)
```

```python
arr3d = np.array([[[1,2],[3,4]],
                  [[5,6],[7,8]]])

# Using for loop
for block in arr3d:
    for row in block:
        for val in row:
            print(val, end=" ")
print()

# Using nditer
for val in np.nditer(arr3d):
    print(val, end=" ")
```

```python
import numpy as np


# Example 2D arrays
arr1 = np.array([[3, 4], [5, 6]])
arr2 = np.array([[1, 0], [7, 8]])

# Average of arrays
avg = (arr1 + arr2) / 2
print("Average:\n", avg)

# Mean
print("Mean:", np.mean([arr1, arr2]))

# Median
print("Median:", np.median([arr1, arr2]))

# Mode
combined = np.concatenate([arr1.flatten(), arr2.flatten()])
print(combined)


```
