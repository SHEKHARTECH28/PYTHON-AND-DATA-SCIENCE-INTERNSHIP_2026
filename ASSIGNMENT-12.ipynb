### 1) Create numpy array and perform following operation

#### Convert 1D array to 2D > Print Array Attributes(Like shape, dimenssion, data type, itemsize)

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])
arr_2d = arr.reshape(2, 3)  
print(arr_2d)
```

```python
print("Shape:", arr_2d.shape)
print("Dimension:", arr_2d.ndim)
print("Data type:", arr_2d.dtype)
print("Item size:", arr_2d.itemsize)
```

 #### Replace all odd numbers in a NumPy array with -1 by itterating using for loop [[23, 56, 78, 93], [71, 82,13, 24]]

```python
arr = np.array([[23, 56, 78, 93],
                [71, 82, 13, 24]])

for i in range(arr.shape[0]):
    for j in range(arr.shape[1]):
        if arr[i, j] % 2 != 0:
            arr[i, j] = -1
print(arr)
```

#### Get the indices of non-zero elements in an array [1, 0, 2, 0, 3, 0, 4]

```python
arr = np.array([1, 0, 2, 0, 3, 0, 4])
indices = np.nonzero(arr)
print(indices)
```

 ### Multiply two NumPy arrays element by element. > Write a code to compute the dot product of two NumPy arrays arr1 = [15, 20, 25] arr2 = [10,40,37]

```python
arr1 = np.array([15, 20, 25])
arr2 = np.array([10, 40, 37])

dot_product = np.dot(arr1, arr2)
print("Dot Product:", dot_product)
```
