```python
#Assignment:- 1

#1) Write a python program that takes in a student name, class. It should also take in five subject marks of
#the students and find the total mark and percentage. 
#Display a result in such a way that their name, class,  and percentage are printed.



Name = str(input("Enter The Name Of Student "))
Class_Stu = input("Enter The Class Of The Student")
Subject1 = float(input("Enter The Marks Of Subject 1"))
Subject2 = float(input("Enter The Marks Of Subject 2"))
Subject3 = float(input("Enter The Marks Of Subject 3"))
Subject4 = float(input("Enter The Marks Of Subject 4"))
Subject5 = float(input("Enter The Marks Of Subject 5"))
Subject6 = float(input("Enter The Marks Of Subject 6"))
total = Subject1+Subject2+Subject3+Subject4+Subject5+Subject6
percentage =total/6
print("\n Result is here")
print("Student Name: ", Name)
print("Class is ",Class_Stu)
print("Total MArks is :" ,total)
print("percentage is :",percentage ,"%")
```

```python

#2) Input 2 strings concatinate them and store in another variable. then perform generally used string methods on it like

#lower(), upper(), title(), swapcase(), capitalize(), casefold(), center(), count(), endswith(), find(), isalnum(),
 #isdigit(), isnumeric(), isspace(), replace()



str1= input("enter your first string: ")
str2= input("enter your second string: ")
str3=str1+" "+str2
print(" Concatenated string :" ,str3)
print("lower():", str3.lower())
print("upper():", str3.upper())
print("title():", str3.title())
print("swapcase():", str3.swapcase())
print("capitalize():", str3.capitalize())
print("casefold():", str3.casefold())
print("center():", str3.center(50, '*'))
print("count('a'):", str3.count('a'))
print("endswith('a'):", str3.endswith('a'))
print("find('a'):", str3.find('a'))
print("isalnum():", str3.isalnum())
print("isdigit():", str3.isdigit())
print("isnumeric():", str3.isnumeric())
print("isspace():", str3.isspace())
print("replace():", str3.replace("a", "@"))
```

```python

#5) In your last program where you find the total and percentage of a student's marks of 5 subject, 
#find the grade of the student using conditional statement.
#Eg. grade 'A' if percentage is greater than or equals to 60, 'B' for  percentage is greater than or equals to 50 and less than 60,
#'C' for  percentage is greater than or equals to 40 and less than 50, 

#'D' for  percentage is greater than or equals to 33 and less than 40, otherwise 'Fail'



Name = str(input("Enter The Name Of Student "))
Class_Stu = input("Enter The Class Of The Student")
Subject1 = float(input("Enter The Marks Of Subject 1"))
Subject2 = float(input("Enter The Marks Of Subject 2"))
Subject3 = float(input("Enter The Marks Of Subject 3"))
Subject4 = float(input("Enter The Marks Of Subject 4"))
Subject5 = float(input("Enter The Marks Of Subject 5"))
Subject6 = float(input("Enter The Marks Of Subject 6"))
total = Subject1+Subject2+Subject3+Subject4+Subject5+Subject6
percentage =total/6
if percentage >= 60:
    grade = 'A'
elif percentage >= 50:
    grade = 'B'
elif percentage >= 40:
    grade = 'C'
elif percentage >= 33:
    grade = 'D'
else:
    grade = 'Fail'
print("\n Result is here")
print("Student Name: ", Name)
print("Class is ",Class_Stu)
print("Total MArks is :" ,total)
print("percentage is :",percentage ,"%")
print("Grade :", grade)
```
