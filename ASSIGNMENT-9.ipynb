### 2) Explore More Datetime Functions in Pandas
### Create Datetime Data

```python
import pandas as pd
data = {
    "Date": ["2026-05-20", "2026-05-21", "2026-05-22"]
}

df = pd.DataFrame(data)

df["Date"] = pd.to_datetime(df["Date"])

print(df)
```

### Extract Date Information

```python
import pandas as pd

data = {
    "Date": ["2026-05-20", "2026-05-21", "2026-05-22"]
}

df = pd.DataFrame(data)

df["Date"] = pd.to_datetime(df["Date"])

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["Day_Name"] = df["Date"].dt.day_name()

print(df)
```

### Date Difference

```python
import pandas as pd

d1 = pd.to_datetime("2026-05-20")
d2 = pd.to_datetime("2026-05-27")

difference = d2 - d1

print(difference.days)
```

### 3) CSV File Data Analysis Using Pandas

### CSV Functions Applied on delhi(air).csv
### load Csv File

```python
import pandas as pd
import numpy as np

# Read CSV file
df = pd.read_csv("delh(air).csv")

# Show dataset
print(df.head())
print(df.tail())

# Dataset information
print(df.info())
print(df.shape)
print(df.columns)

# Check missing values
print(df.isnull().sum())

# Data cleaning
df.drop_duplicates(inplace=True)
df.fillna(0, inplace=True)

# Convert datetime
df["DateTime"] = pd.to_datetime(df["DateTime"])

# Extract date details
df["Year"] = df["DateTime"].dt.year
df["Month"] = df["DateTime"].dt.month
df["Day"] = df["DateTime"].dt.day
df["Hour"] = df["DateTime"].dt.hour

# Show extracted columns
print(df[["DateTime", "Year", "Month", "Day", "Hour"]].head())

# Statistical summary
print(df.describe())

# Average pollution
avg_pollution = df["Values"].mean()
print("Average Pollution :", avg_pollution)

# Maximum pollution
max_pollution = df["Values"].max()
print("Maximum Pollution :", max_pollution)

# Minimum pollution
min_pollution = df["Values"].min()
print("Minimum Pollution :", min_pollution)

# Most polluted location
most_polluted = df.groupby("Locations")["Values"].mean().sort_values(ascending=False)
print(most_polluted.head(1))

# Least polluted location
least_polluted = df.groupby("Locations")["Values"].mean().sort_values()
print(least_polluted.head(1))



# Top 10 pollution records
top10 = df.sort_values(by="Values", ascending=False).head(10)
print(top10)

# Save cleaned dataset
df.to_csv("cleaned_delhi_air_data.csv", index=False)

print("Dataset cleaned and saved successfully")
```
