### 2)Create two DataFrames, df1 and df2, with a common column (e.g., 'ID'). 


```python
import pandas as pd 
df1=pd.DataFrame({
    'Name':['A1','A2','A3','A4','A5'],
    'subject_id':['sub1','sub2','sub3','sub4','sub5'], 
'Marks_scored':[98,90,87,69,78],
'ID':[101,102,103,104,105]
})
df2=pd.DataFrame({
    'Name': ['B1', 'B2', 'B3', 'B4', 'B5'], 
'subject_id':['sub2','sub4','sub3','sub6','sub5'], 
'Marks_scored':[89,80,79,97,88],
'ID':[101,102,103,104,105]})

```

### >Perform an inner merge on this common column and display the resulting DataFrame

```python
import pandas as pd 
df1=pd.DataFrame({
    'Name':['A1','A2','A3','A4','A5'],
    'subject_id':['sub1','sub2','sub3','sub4','sub5'], 
'Marks_scored':[98,90,87,69,78],
'ID':[101,102,103,104,105]
})
df2=pd.DataFrame({
    'Name': ['B1', 'B2', 'B3', 'B4', 'B5'], 
'subject_id':['sub2','sub4','sub3','sub6','sub5'], 
'Marks_scored':[89,80,79,97,88],
'ID':[101,102,103,104,105]})

print(df1.merge(df2, on='ID', how='inner')) 
```

### 3) Create three DataFrames

```python
import pandas as pd 
df1=pd.DataFrame({
    'Name':['A1','A2','A3','A4','A5'],
    'subject_id':['sub1','sub2','sub3','sub4','sub5'], 
'Marks_scored':[98,90,87,69,78],
'ID':[101,102,103,104,105]
})
df2=pd.DataFrame({
    'Name': ['B1', 'B2', 'B3', 'B4', 'B5'], 
'subject_id':['sub2','sub4','sub3','sub6','sub5'], 
'Marks_scored':[89,80,79,97,88],
'ID':[101,102,103,104,105]})
df3=pd.DataFrame({
    'Name': ['C1', 'C2', 'C3', 'C4', 'C5'], 
'subject_id':['sub4','sub3','sub2','sub1','sub5'], 
'Marks_scored':[79,85,71,93,90],
'ID':[101,102,103,104,105]})

```

###  Vertically concatenate two of them using pd.concat()

```python
result=pd.concat([df1,df2])
print(result)
```

### then merge the resulting DataFrame with the third DataFrame on a common key

```python
res1=pd.merge(result,df3,on='ID',how='inner')
print(res1)
```

### Also Explain the primary differences between df.join() and pd.merge(),
pd.merge()
   - Used to combine DataFrames using columns or indexes.
   - More flexible.
   - Similar to SQL joins.

2. df.join()
   - Used mainly for joining on indexes.
   - Simpler and faster for index-based joining.

3. merge() supports:
   - inner join
   - left join
   - right join
   - outer join
   - multiple keys

4. join() is mostly used when indexes are already set.
